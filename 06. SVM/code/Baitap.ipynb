## Bài 1


```python
# -*- coding: utf-8 -*-
"""Bài 1: tính tay khoảng cách tới ranh giới."""

import numpy as np
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

X, y = make_blobs(
    n_samples=24, centers=[[2, 2], [6, 6]], cluster_std=0.9, random_state=0
)
y = 1 - y
svm = SVC(kernel="linear", C=1000).fit(X, y)
w = svm.coef_[0]
b = svm.intercept_[0]
do_dai_w = np.sqrt(w[0] ** 2 + w[1] ** 2)

print(f"w1 = {w[0]:.6f}, w2 = {w[1]:.6f}, b = {b:.6f}")
print(f"||w|| = {do_dai_w:.6f}")
print("\nHai diem can tinh:")
for diem in np.array([[3.0, 3.0], [5.0, 4.5]]):
    f = w @ diem + b
    khoang_cach = abs(f) / do_dai_w
    print(f"  {tuple(diem)}: f(x) = {f:+.6f}, khoang cach = {khoang_cach:.6f}")

print("\nBa vecto ho tro:")
for i in svm.support_:
    f = w @ X[i] + b
    khoang_cach = abs(f) / do_dai_w
    print(f"  diem {i:2d} {tuple(X[i].round(4))}: f(x) = {f:+.6f}, khoang cach = {khoang_cach:.6f}")

print("\nNhan xet:")
print(f"- Moi vecto ho tro cach ranh gioi xap xi 1/||w|| = {1/do_dai_w:.6f}.")
print("- Ly do: vecto ho tro nam tren hai mep le f(x)=+1 hoac f(x)=-1,")
print("  nen |f(x)|/||w|| = 1/||w||. Hai mep le cach nhau 2/||w||.")
```

## Bài 2


```python
# -*- coding: utf-8 -*-
"""Bài 2: thêm một điểm và giải thích bằng khái niệm vectơ hỗ trợ."""

import numpy as np
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

X, y = make_blobs(
    n_samples=24, centers=[[2, 2], [6, 6]], cluster_std=0.9, random_state=0
)
y = 1 - y

def hoc_va_in(ten, X_hoc, y_hoc):
    mh = SVC(kernel="linear", C=1000).fit(X_hoc, y_hoc)
    w = mh.coef_[0]
    b = mh.intercept_[0]
    print(f"{ten}")
    print(f"  w = ({w[0]:+.6f}, {w[1]:+.6f}), b = {b:+.6f}")
    print(f"  chi so vecto ho tro = {mh.support_.tolist()}")
    return mh

goc = hoc_va_in("Mo hinh goc", X, y)

X_xa = np.vstack([X, [-1.0, -1.0]])
y_xa = np.append(y, 1)
them_xa = hoc_va_in("\nThem diem lop 1 tai (-1, -1)", X_xa, y_xa)
f_xa = them_xa.decision_function([[-1.0, -1.0]])[0]
print(f"  f(diem moi) = {f_xa:+.6f}; diem moi la vecto ho tro: {24 in them_xa.support_}")

X_gan = np.vstack([X, [4.5, 4.0]])
y_gan = np.append(y, 1)
them_gan = hoc_va_in("\nThay bang diem lop 1 tai (4.5, 4)", X_gan, y_gan)
f_gan = them_gan.decision_function([[4.5, 4.0]])[0]
print(f"  f(diem moi) = {f_gan:+.6f}; diem moi la vecto ho tro: {24 in them_gan.support_}")

print("\nMuc thay doi so voi mo hinh goc:")
for ten, mh in [("(-1, -1)", them_xa), ("(4.5, 4)", them_gan)]:
    print(f"  {ten:10s}: ||delta w|| = {np.linalg.norm(mh.coef_[0]-goc.coef_[0]):.6f}, "
          f"|delta b| = {abs(mh.intercept_[0]-goc.intercept_[0]):.6f}")

print("\nNhan xet:")
print("- Diem (-1,-1) nam xa ranh gioi ve dung phia lop 1, khong thanh vecto ho tro;")
print("  vi vay duong ranh gioi gan nhu khong thay doi.")
print("- Diem (4.5,4) nam sat/vuong vao le, tro thanh vecto ho tro va lam thay doi ranh gioi.")
print("- SVM duoc quyet dinh boi cac diem gan le, khong phai tat ca diem hoc.")
```

## Bài 3


```python
# -*- coding: utf-8 -*-
"""Bài 3: dò C cho kernel đa thức bậc 2, 3 và 4."""

import pandas as pd
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

df = pd.read_csv("data/hai_vang_trang.csv")
X = df[["x1", "x2"]]
y = df["nhan"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

ket_qua = []
for bac in [2, 3, 4]:
    pipe = make_pipeline(StandardScaler(), SVC(kernel="poly", degree=bac))
    tim = GridSearchCV(pipe, {"svc__C": [0.1, 1, 10, 100]}, cv=5)
    tim.fit(X_tr, y_tr)
    ket_qua.append((bac, tim.best_params_["svc__C"], tim.best_score_, tim.score(X_te, y_te)))

print("bac  C tot nhat  diem CV  accuracy kiem tra")
for bac, C, cv, test in ket_qua:
    print(f" {bac:1d}      {C:6g}    {cv:.4f}       {test:.4f}")

tot_nhat = max(ket_qua, key=lambda dong: dong[2])
print("\nNhan xet:")
print(f"- Theo diem kiem dinh cheo, kernel da thuc tot nhat la bac {tot_nhat[0]} "
      f"voi C={tot_nhat[1]:g}, diem CV={tot_nhat[2]:.4f}.")
print(f"- Accuracy kiem tra cua cau hinh nay la {tot_nhat[3]:.4f}.")
print("- Kernel RBF o muc 5 dat 0.9200 tren tap kiem tra; doi chieu bang tren de ket luan")
print("  kernel da thuc co vuot RBF hay khong. Chon mo hinh dua tren CV, khong dua tren tap kiem tra.")
```

## Bài 4


```python
# -*- coding: utf-8 -*-
"""Bài 4: dò lưới mịn quanh C=10, gamma=1."""

import pandas as pd
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.svm import SVC

df = pd.read_csv("data/hai_vang_trang.csv")
X = df[["x1", "x2"]]
y = df["nhan"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

luoi = {"C": [3, 10, 30], "gamma": [0.3, 1, 3]}
tim = GridSearchCV(SVC(kernel="rbf"), luoi, cv=5)
tim.fit(X_tr, y_tr)
bang = pd.DataFrame(tim.cv_results_["params"])
bang["diem"] = tim.cv_results_["mean_test_score"]
print("Diem kiem dinh cheo cua luoi min:")
print(bang.pivot(index="C", columns="gamma", values="diem").round(4).to_string())
print("\nCap tot nhat:", tim.best_params_)
print(f"Diem CV tot nhat: {tim.best_score_:.4f}")
print(f"Accuracy tap kiem tra: {tim.score(X_te, y_te):.4f}")

print("\nNhan xet:")
print(f"- Diem CV tot nhat {'cao hon' if tim.best_score_ > 0.8711 else 'khong cao hon'} 0.8711 cua luoi tho.")
print(f"- Accuracy kiem tra {'doi' if abs(tim.score(X_te,y_te)-0.9200)>1e-12 else 'khong doi'}, "
      f"so voi 0.9200 cua luoi tho.")
print("- Luoi min giup kiem tra ky vung tham so tot, nhung khong bao dam tap kiem tra se tang.")
```

## Bài 5


```python
# -*- coding: utf-8 -*-
"""Bài 5: bảng nhầm lẫn và trực quan các ảnh SVM đoán sai."""

from collections import Counter
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_digits
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X, y = load_digits(return_X_y=True)
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)
svm = SVC(kernel="rbf").fit(X_tr, y_tr)
du_doan = svm.predict(X_te)
cm = confusion_matrix(y_te, du_doan)
print("Bang nham lan (hang=chu so that, cot=chu so doan):")
print(cm)

sai = np.flatnonzero(du_doan != y_te)
cap_sai = Counter(zip(y_te[sai], du_doan[sai]))
print("\nCac cap bi nham:")
for (that, doan), so_lan in cap_sai.most_common():
    print(f"  {that} -> {doan}: {so_lan} lan")

so_anh = len(sai)
fig, axes = plt.subplots(1, so_anh, figsize=(3 * so_anh, 3))
axes = np.atleast_1d(axes)
for ax, i in zip(axes, sai):
    ax.imshow(X_te[i].reshape(8, 8), cmap="gray_r")
    ax.set_title(f"That {y_te[i]}, doan {du_doan[i]}")
    ax.axis("off")
fig.tight_layout()
Path("figures").mkdir(exist_ok=True)
fig.savefig("figures/bai5_chu_so_doan_sai.png", dpi=160, bbox_inches="tight")
plt.show()

print("\nNhan xet:")
print("- Mo hinh chi sai o mot vai chu so co net viet mo, dut hoac hinh dang gan voi lop khac.")
print("- Vi du 9 co the giong 3 khi net vong ben trai mo; 8 co the giong 1 neu hai vong bi dut;")
print("  4 co the giong 8 khi cac net khep kin; 7 co the giong 9 khi co them net cong.")
print("- Da luu hinh cac anh sai tai figures/bai5_chu_so_doan_sai.png.")
```

## Bài 6


```python
# -*- coding: utf-8 -*-
"""Bài 6: đo thời gian SVC và LinearSVC khi số mẫu tăng."""

from time import perf_counter
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.svm import LinearSVC, SVC

dong = []
for n in [2000, 5000, 10000, 20000]:
    X, y = make_classification(
        n_samples=n, n_features=20, random_state=0
    )
    for ten, mh in [
        ("SVC", SVC()),
        ("LinearSVC", LinearSVC(dual="auto", max_iter=10000, random_state=0)),
    ]:
        bat_dau = perf_counter()
        mh.fit(X, y)
        thoi_gian = perf_counter() - bat_dau
        dong.append({"n": n, "mo_hinh": ten, "giay": thoi_gian})

bang = pd.DataFrame(dong)
pivot = bang.pivot(index="n", columns="mo_hinh", values="giay")
print("Thoi gian fit (giay):")
print(pivot.round(4).to_string())

print("\nHe so tang thoi gian khi n tang:")
for ten in ["SVC", "LinearSVC"]:
    ds = pivot[ten].to_numpy()
    print(ten, [round(ds[i] / ds[i-1], 2) for i in range(1, len(ds))])

print("\nNhan xet:")
print("- LinearSVC tang gan tuyen tinh theo so mau va phu hop hon cho du lieu lon.")
print("- SVC kernel RBF phai tinh quan he giua nhieu cap diem, nen thoi gian thuong tang nhanh")
print("  hon tuyen tinh; khi n tang gap doi, thoi gian co the tang nhieu hon gap doi.")
print("- So lieu thuc te phu thuoc cau hinh may, nhung xu huong ve do phuc tap van giu nguyen.")
```
